# 第1回 四則演算

式が書かれた枠を「コードセル」と呼びます。コードセルをクリックし、`Shift` キーを押しながら `Enter` キーを押すと、計算結果が下に表示されます。

数値や記号は半角で入力してください。式を書き換えて、同じセルを何度でも実行できます。

## 四則演算

Pythonでは、足し算に `+`、引き算に `-`、掛け算に `*`、割り算に `/` を使います。次の4つのセルを上から順に実行してください。

```python
8 + 3
```

```python
8 - 3
```

```python
8 * 3
```

```python
8 / 4
```

## 小数の計算

`/` による割り算では、割り切れる場合も `2.0` のように小数で結果が表示されます。小数を入力するときは、小数点に `.` を使います。

```python
1.5 + 2.25
```

## 計算の順序と括弧

掛け算と割り算は、足し算と引き算より先に計算されます。先に計算したい部分を `( )` で囲むと、計算の順序を指定できます。次の2つの結果を比べてください。

```python
2 + 3 * 4
```

```python
(2 + 3) * 4
```
